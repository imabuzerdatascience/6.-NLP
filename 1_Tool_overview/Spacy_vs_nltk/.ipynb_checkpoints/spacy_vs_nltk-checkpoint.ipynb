<h1 align="center" >Sentences and word tokennization using spacy </h1> 

```python
import spacy 
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
docs = nlp("Dr. Abuzer is coming india and eat pani puri . Mr Abhinash is seelping.")
```

```python
for sentences in docs.sents :
    print(sentences)
```

```python
for sentences in docs.sents :
    for words in sentences :
        print(words)
```

```python
Eassy_docs = nlp("The cow . is a highly valued, gentle domestic animal kept worldwide for its milk, which is rich in calcium and protein. Primarily herbivores, cows graze on grass and fodder, contributing to agriculture through natural fertilizer (dung) and labor. They are considered sacred in some cultures and are vital to dairy industries.")
```

```python
for eassy_sentences in Eassy_docs.sents :
    print(eassy_sentences)
```

```python
for eassy_sentences in Eassy_docs.sents :
    for word in eassy_sentences :
        print(word)
```

<h1 align="center" >Sentences and word tokennization using NLTK </h1> 

```python
import nltk 
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize("Dr. Abuzer is coming india and eat pani puri . Mr Abhinash is seelping.")
```

```python
from nltk.tokenize import word_tokenize
word_tokenize("Dr. Abuzer is coming india and eat pani puri . Mr Abhinash is seelping.")
```

```python
sent_tokenize("The cow . is a highly valued, gentle domestic animal kept worldwide for its milk, which is rich in calcium and protein. Primarily herbivores, cows graze on grass and fodder, contributing to agriculture through natural fertilizer (dung) and labor. They are considered sacred in some cultures and are vital to dairy industries.")
```
