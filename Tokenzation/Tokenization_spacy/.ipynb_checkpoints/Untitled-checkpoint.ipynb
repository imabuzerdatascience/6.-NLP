```python
import spacy 
```

```python
nlp = spacy.blank("en")
doc = nlp("The Marvel Cinematic Universe (MCU) features over 30 interconnected superhero films, produced by Marvel Studios and distributed by Disney")
```

```python
for word in doc :
    print(word)
```

```python
for token in doc :
    print ( token , "==>" , "index" , token.i , "is_alpha" , token.is_alpha ,
            "is_punct" , token.is_punct ,
            "like_num " , token.like_num , 
            "is_currency" , token.is_currency )
            
```
