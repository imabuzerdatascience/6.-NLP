```python
import spacy 
```

```python
nlp = spacy.blank("en")
doc = nlp("The Marvel Cinematic Universe (MCU) features over 30 interconnected superhero films, produced by Marvel Studios and distributed by Disney")
```

```python
for word in doc :
    print(word)
```
